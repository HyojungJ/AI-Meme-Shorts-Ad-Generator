```python
import pandas as pd
import json
from datetime import datetime
```

```python
df = pd.read_csv("smilestyle_dataset.tsv", sep="\t")

gentle_data = df['gentle'].dropna().tolist()
gentle_data = [text for text in gentle_data if text.strip()]
```

```python
# 텍스트 파일로 저장
with open("gentle_data.txt", 'w', encoding='utf-8') as f:
    for text in gentle_data:
        f.write(f"{text}\n")

# JSON 파일로 저장
json_data = {
    "data": gentle_data
}

with open("gentle_data.json", 'w', encoding='utf-8') as f:
    json.dump(json_data, f, ensure_ascii=False, indent=2)
```
