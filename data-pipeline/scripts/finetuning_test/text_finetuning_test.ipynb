### Llama 3

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig

model_id = "meta-llama/Llama-3.1-8B-Instruct"

# 4비트 양자화 설정
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.bfloat16
)

tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    quantization_config=bnb_config,
    device_map="auto",
    low_cpu_mem_usage=True
)
print("Llama 3 모델 로드 완료")
```

```python
def test_character(persona_type, situation, action):
    # 규칙 프롬프트
    if persona_type == "A":
        system_prompt = "너는 에너지가 넘치는 인플루언서야. 문장 끝에 '!!'를 자주 쓰고, '대박', '진짜' 같은 감탄사를 맨 앞에 써줘. 말투는 '~했어?', '~인정?' 처럼 친근하게 해줘."
    else:
        system_prompt = "너는 다정하고 차분한 인플루언서야. '포근한', '소중한' 같은 단어를 쓰고, 말투는 '~요', '~네요'로 끝나는 감성적인 스타일로 해줘."

    # Llama 3 전용 포맷 적용
    prompt = f"""<|begin_of_text|><|start_header_id|>system<|end_header_id|>
                {system_prompt}<|eot_id|><|start_header_id|>user<|end_header_id|>
                상황: {situation}, 지문: [{action}]<|eot_id|><|start_header_id|>assistant<|end_header_id|>
              """

    inputs = tokenizer(prompt, return_tensors="pt").to("cuda")
    outputs = model.generate(**inputs, max_new_tokens=100, do_sample=True, temperature=0.7, eos_token_id=tokenizer.eos_token_id)

    result = tokenizer.decode(outputs[0], skip_special_tokens=True)
    # 프롬프트 부분 제외하고 답변만 출력
    print(f"Llama 3 모델 답변:\n{result.split('assistant')[-1].strip()}")
```

```python
# 테스트 실행
for data in test_data:
    print(f"\n검증 캐릭터 유형: {data['type']}")
    print(f"입력 상황: {data['situation']}")
    print(f"입력 지문: {data['action']}")
    print("-" * 30)

    # 모델 실행
    test_character(data['type'], data['situation'], data['action'])

    print(f"기대한 정답(Target):\n{data['target']}")
    print("=" * 50)
```

### EEVE

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig

# EEVE 모델
model_id = "yanolja/YanoljaNEXT-EEVE-Instruct-10.8B"

# 4비트 양자화 설정
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.bfloat16
)

tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    quantization_config=bnb_config,
    device_map="auto"
)
print("EEVE 모델 로드 완료")
```

```python
def test_eeve(persona_type, situation, action):
    if persona_type == "A":
        system_prompt = "너는 에너지가 넘치는 인플루언서야. 문장 끝에 '!!'를 자주 쓰고, '대박', '진짜' 같은 감탄사를 맨 앞에 써줘. 말투는 '~했어?', '~인정?' 처럼 친근하게 해줘."
    else:
        system_prompt = "너는 다정하고 차분한 인플루언서야. '포근한', '소중한' 같은 단어를 쓰고, 말투는 '~요', '~네요'로 끝나는 감성적인 스타일로 해줘."

    # EEVE 전용 ChatML 포맷
    prompt = f"""<|im_start|>system
                {system_prompt}<|im_end|>
                <|im_start|>user
                상황: {situation}, 지문: [{action}]<|im_end|>
                <|im_start|>assistant
              """

    inputs = tokenizer(prompt, return_tensors="pt").to("cuda")
    outputs = model.generate(
        **inputs,
        max_new_tokens=100,
        do_sample=True,
        temperature=0.8,
        top_p=0.9,
        eos_token_id=tokenizer.eos_token_id
    )

    result = tokenizer.decode(outputs[0], skip_special_tokens=True)
    print(f"EEVE 답변:\n{result.split('assistant')[-1].strip()}")
```

```python
for data in test_data:
    print(f"\n검증 캐릭터 유형: {data['type']}")
    print(f"입력 상황: {data['situation']}")
    print(f"입력 지문: {data['action']}")
    print("-" * 30)

    # EEVE 전용 함수 호출
    test_eeve(data['type'], data['situation'], data['action'])

    print(f"기대한 정답(Target):\n{data['target']}")
    print("=" * 50)
```

### Solar

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig

# Solar 한국어 특화 모델
model_id = "upstage/SOLAR-10.7B-Instruct-v1.0"

# 4비트 양자화 설정
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.bfloat16
)

tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    quantization_config=bnb_config,
    device_map="auto"
)
print("Solar 모델 로드 완료!")
```

```python
def test_solar(persona_type, situation, action):
    if persona_type == "A":
        sys_msg = "에너지가 넘치는 한국인 인플루언서야. 문장 끝에 '!!'를 쓰고, '대박', '진짜'를 써줘. 친근한 반말."
    else:
        sys_msg = "다정하고 차분한 한국인 인플루언서야. '포근한', '소중한' 단어를 쓰고, 감성적인 말투. 존댓말."

    # Solar 공식 포맷
    prompt = f"<s> [INST] {sys_msg}\n상황: {situation}\n지문: {action} [/INST]"

    inputs = tokenizer(prompt, return_tensors="pt").to("cuda")
    
    with torch.no_grad():
        outputs = model.generate(
            **inputs, 
            max_new_tokens=150,
            do_sample=True,
            temperature=0.8, 
            top_p=0.9,
            repetition_penalty=1.1,
            # 생성 중단 방지
            pad_token_id=tokenizer.eos_token_id 
        )
    
    # 1. 전체 텍스트 디코딩
    full_text = tokenizer.decode(outputs[0], skip_special_tokens=False)
    
    # 2. [/INST] 태그 이후의 내용만 
    if "[/INST]" in full_text:
        answer = full_text.split("[/INST]")[-1].replace("</s>", "").strip()
    else:
        # 태그가 안 보일 경우를 대비해 아예 전체 출력 시도
        answer = tokenizer.decode(outputs[0][inputs.input_ids.shape[1]:], skip_special_tokens=True).strip()
    
    # 3. 결과 출력
    if not answer:
        # 만약 그래도 비어있다면, 모델이 생성한 원본 토큰 자체를 강제로 확인
        print(f"⚠️ 원본 데이터 확인용: {full_text}") 
    else:
        print(f"🤖 Solar 답변:\n{answer}")
```

```python
for data in test_data:
    print(f"\n검증 캐릭터 유형: {data['type']}")
    print(f"입력 상황: {data['situation']}")
    print(f"입력 지문: {data['action']}")
    print("-" * 30)

    # Solar 전용 함수 호출
    test_solar(data['type'], data['situation'], data['action'])

    print(f"기대한 정답(Target):\n{data['target']}")
    print("=" * 50)
```
