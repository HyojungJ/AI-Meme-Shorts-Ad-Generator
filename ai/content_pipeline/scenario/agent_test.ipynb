```python
from content_pipeline.scenario.prompts.generate_scenario_templates import GENERATE_SCENARIO_TEMPLATE_V4
from content_pipeline.scenario.prompts.meme_example_templates import MEME_EXAMPLE_TEMPLATE_V1
from content_pipeline.scenario.prompts.regenerate_scenario_templates import REGENERATE_SCENARIO_TEMPLATE_V3
from content_pipeline.scenario.prompts.review_scenario_templates import REVIEW_SCENARIO_TEMPLATE_V3
from content_pipeline.scenario.scenario_agent import run_scenario_agent
from content_pipeline.scenario.scenario_database import save_prompt_template_to_db

from dotenv import load_dotenv
load_dotenv()

# 템플릿 저장
save_prompt_template_to_db("GENERATE_SCENARIO_TEMPLATE_V4", "4", GENERATE_SCENARIO_TEMPLATE_V4)
save_prompt_template_to_db("MEME_EXAMPLE_TEMPLATE_V1", "1", MEME_EXAMPLE_TEMPLATE_V1)
save_prompt_template_to_db("REGENERATE_SCENARIO_TEMPLATE_V3", "3", REGENERATE_SCENARIO_TEMPLATE_V3)
save_prompt_template_to_db("REVIEW_SCENARIO_TEMPLATE_V3", "3", REVIEW_SCENARIO_TEMPLATE_V3)


# 시나리오 생성
result = run_scenario_agent(
    ad_id=17, 
    generate_template=GENERATE_SCENARIO_TEMPLATE_V4,
    review_template=REVIEW_SCENARIO_TEMPLATE_V3,
    regenerate_template=REGENERATE_SCENARIO_TEMPLATE_V3,
    meme_example_template=MEME_EXAMPLE_TEMPLATE_V1,
    used_templates=["GENERATE_SCENARIO_TEMPLATE_V4",
                    "REVIEW_SCENARIO_TEMPLATE_V3",
                    "REGENERATE_SCENARIO_TEMPLATE_V3",
                    "MEME_EXAMPLE_TEMPLATE_V1"]
)

if result["status"] == "completed":
    print(f"Script ID: {result['script_id']}")
```
